```python
import pandas as pd
from dotenv import load_dotenv
import os

load_dotenv()

from sqlalchemy import create_engine



```

```python
df = pd.read_csv("amazon_india_2015_to_2025_cleaned.csv")
```

```python

user = os.getenv("DB_USER")
password = os.getenv("DB_PASSWORD")
host = os.getenv("DB_HOST")
db = os.getenv("DB_NAME")
```

```python
# Transactions table
transactions = df[[
    "transaction_id",
    "order_date",
    "customer_id",
    "product_id",
    "quantity",
    "original_price_inr",  
    "subtotal_inr",
    "delivery_charges",
    "final_amount_inr",
    "discount_percent",
    "payment_method",
    "delivery_days",
    "delivery_type",
    "is_prime_member",
    "is_festival_sale",
    "customer_rating",
    "return_status"
]]


# Products table
products = df[[
    "product_id","product_name","category","subcategory","brand",
    "original_price_inr","discounted_price_inr",
    "product_weight_kg","product_rating","is_prime_eligible"
]].drop_duplicates()

# Customers table
customers = df[[
    "customer_id","customer_city","customer_state",
    "customer_tier","customer_spending_tier","customer_age_group"
]].drop_duplicates()

# Time dimension
time_dim = df[["order_date","order_month","order_year","order_quarter"]]
time_dim = time_dim.drop_duplicates()
```

```python

engine = create_engine(
    f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:3306/{DB_NAME}"
)


products.to_sql("products", engine, if_exists="replace", index=False)
customers.to_sql("customers", engine, if_exists="replace", index=False)
time_dim.to_sql("time_dimension", engine, if_exists="replace", index=False)
transactions.to_sql("transactions", engine, if_exists="replace", index=False)
```

```python
engine = create_engine(
    f"postgresql+psycopg2://postgres:{DB_PASSWORD}@l{DB_HOST}:5432/{DB_NAME}"
)

products.to_sql("products", engine, if_exists="replace", index=False)
customers.to_sql("customers", engine, if_exists="replace", index=False)
time_dim.to_sql("time_dimension", engine, if_exists="replace", index=False)
transactions.to_sql("transactions", engine, if_exists="replace", index=False)
```
